Splitting of data! 80-20

```python
import pandas as pd

result = pd.read_csv("streaming_movies_tv_merged.csv")
print(result.shape)
```

```python
from sklearn.model_selection import train_test_split

explore, holdout = train_test_split(result, test_size=0.20, random_state=226)
print(len(explore), len(holdout))

explore.to_csv("/content/explore_80.csv", index=False) #index=False stops pandas from adding an extra row-number column.
holdout.to_csv("/content/holdout_20.csv", index=False)
```
